```python
TIERS = {
    "unacceptable": {
        "tier": "UNACCEPTABLE RISK",
        "article": "Article 5",
        "legal_basis": "Prohibited entirely under EU AI Act.",
        "obligations": "Cannot be placed on the market or put into service in the EU.",
        "principle": "Violates human autonomy, dignity, or fundamental rights."
    },
    "high": {
        "tier": "HIGH RISK",
        "article": "Article 6 + Annex III",
        "legal_basis": "Permitted but subject to strict compliance obligations.",
        "obligations": "Requires conformity assessment, risk management system, human oversight, and EU registration.",
        "principle": "Significant potential impact on health, safety, or fundamental rights."
    },
    "limited": {
        "tier": "LIMITED RISK",
        "article": "Article 50",
        "legal_basis": "Permitted with transparency obligations only.",
        "obligations": "Must disclose to users that they are interacting with an AI system.",
        "principle": "Risk of manipulation or deception without disclosure."
    },
    "minimal": {
        "tier": "MINIMAL RISK",
        "article": "Recital 48",
        "legal_basis": "Permitted. No obligations under EU AI Act.",
        "obligations": "No mandatory requirements. Voluntary codes of conduct encouraged.",
        "principle": "Negligible risk to fundamental rights or public safety."
    }
}

print("Tier definitions loaded.")
```

```python
# ── KEYWORDS ─────────────────────────────────────────────────────

UNACCEPTABLE_KEYWORDS = [
    "subliminal",
    "manipulate",
    "deceive",
    "social scor",
    "citizen score",
    "scores citizens",
    "social behaviour",
    "predict crime",
    "predict criminal",
    "criminal profil",
    "scrape face",
    "facial database",
    "biometric surveillance",
    "emotion detection",
    "infer race",
    "infer religion",
    "infer sexuality",
    "real-time facial",
    "real time biometric"
]

HIGH_RISK_KEYWORDS = [
    "credit scor",
    "loan decision",
    "recruitment",
    "cv screening",
    "hiring decision",
    "job application",
    "screens cv",
    "screen cv",
    "cv screen",
    "biometric identification",
    "border control",
    "critical infrastructure",
    "water supply",
    "energy grid",
    "education assessment",
    "student evaluation",
    "medical device",
    "clinical decision",
    "patient diagnosis",
    "law enforcement",
    "polygraph",
    "insurance risk",
    "autonomous vehicle",
    "self driving",
    "worker monitoring",
    "employment decision"
]

LIMITED_RISK_KEYWORDS = [
    "chatbot",
    "virtual assistant",
    "deepfake",
    "synthetic media",
    "ai generated content",
    "emotion recognition",
    "sentiment analysis",
    "recommendation system"
]

# ── CLASSIFICATION FUNCTION ───────────────────────────────────────

def classify_ai_use_case(description):
    description = description.lower()

    for keyword in UNACCEPTABLE_KEYWORDS:
        if keyword in description:
            return TIERS["unacceptable"]

    for keyword in HIGH_RISK_KEYWORDS:
        if keyword in description:
            return TIERS["high"]

    for keyword in LIMITED_RISK_KEYWORDS:
        if keyword in description:
            return TIERS["limited"]

    return TIERS["minimal"]

print("Keywords and classifier loaded.")
```

```python
# ── TEST YOUR OWN CASES ───────────────────────────────────────────

def test(description):
    result = classify_ai_use_case(description)
    print("\n" + "="*60)
    print(f"INPUT:          {description}")
    print(f"CLASSIFICATION: {result['tier']}")
    print(f"Article:        {result['article']}")
    print(f"Legal Basis:    {result['legal_basis']}")
    print(f"Obligations:    {result['obligations']}")
    print(f"Principle:      {result['principle']}")
    print("-"*60)
    print("DISCLAIMER: This classification is indicative only.")
    print("Final determination requires qualified legal review")
    print("under EUR-Lex 2024/1689 (EU AI Act).")
    print("="*60)

# ── CHANGE THESE TO TEST ANYTHING YOU WANT ───────────────────────

test("A system that scores citizens based on social behaviour")
test("An AI that screens CVs for job applications")
test("A customer service chatbot")
test("A spam filter for emails")
```
